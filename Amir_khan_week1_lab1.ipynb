```python
# Q2: Check if a number is prime by testing its divisibility.

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


print(is_prime(7))
print(is_prime(10))
print(is_prime(2))
```

```python
# Q3: Print the Fibonacci sequence up to n terms using a loop.

n = 10

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
# Q4: Remove duplicates from a list while preserving the original order.

def remove_duplicates(numbers):
    new_list = []

    for number in numbers:
        if number not in new_list:
            new_list.append(number)

    return new_list


numbers = [1, 2, 2, 3, 4, 3, 5, 1]

print(remove_duplicates(numbers))
```

```python
# Q5: Count vowels, consonants, digits, and spaces in a string.

text = "Hello World 123"

vowels = 0
consonants = 0
digits = 0
spaces = 0

for char in text:
    if char.lower() in "aeiou":
        vowels += 1
    elif char.isalpha():
        consonants += 1
    elif char.isdigit():
        digits += 1
    elif char == " ":
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
```

```python
# Q6: Multiply any number of arguments using *args.

def multiply(*args):
    product = 1

    for number in args:
        product *= number

    return product


print(multiply(2, 3, 4))
print(multiply(5, 6))
```

```python
#7 Approach: Use a dictionary comprehension to count each unique character in the string.

text = "hello"

frequency = {char: text.count(char) for char in set(text)}

print(frequency)
```

```python
# Q17
# Approach: Find the minimum and maximum values of the array,
# then apply the normalization formula (x - min) / (max - min).

import numpy as np

arr = np.array([10, 20, 30, 40])

normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print("Original array:", arr)
print("Normalized array:", normalized)
```

```python
# Q20
# Approach: Create 16 numbers using np.arange(), reshape them into
# a 4x4 matrix, then compute its transpose, determinant, and inverse.

import numpy as np

matrix = np.arange(1, 17).reshape(4, 4)

transpose = matrix.T
determinant = np.linalg.det(matrix)

print("Matrix:")
print(matrix)

print("\nTranspose:")
print(transpose)

print("\nDeterminant:")
print(determinant)

if determinant != 0:
    inverse = np.linalg.inv(matrix)
    print("\nInverse:")
    print(inverse)
else:
    print("\nMatrix is not invertible.")
```

```python
# Q21
# Approach: Calculate the dot product manually by multiplying
# corresponding elements in a loop and adding them. Then verify
# the result using np.dot().

import numpy as np

A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

dot_product = 0

for i in range(len(A)):
    dot_product += A[i] * B[i]

print("Manual dot product:", dot_product)
print("NumPy dot product:", np.dot(A, B))
```

```python
# Q24
# Approach: Compare the array with its transpose.
# If all values are equal, return True; otherwise return False.

import numpy as np

def is_symmetric(arr):
    return (arr == arr.T).all()


A = np.array([
    [1, 2, 3],
    [2, 4, 5],
    [3, 5, 6]
])

print("Is A symmetric?", is_symmetric(A))
```

```python
# Q25
# Approach: Create three 2x2 arrays, then stack them vertically
# using np.vstack() and horizontally using np.hstack().
# Finally, print their shapes to see the difference.

import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

C = np.array([
    [9, 10],
    [11, 12]
])

vertical = np.vstack((A, B, C))
horizontal = np.hstack((A, B, C))

print("Vertical stacking:")
print(vertical)
print("Shape:", vertical.shape)

print()

print("Horizontal stacking:")
print(horizontal)
print("Shape:", horizontal.shape)

#Difference b/w np.vstack() and np.hstack()
# np.vstack() stacks arrays vertically, increasing the number of rows.
# np.hstack() stacks arrays horizontally, increasing the number of columns.
```

```python
# Q27
# Approach: Create 100 evenly spaced x values from 0 to 2π,
# calculate sin(x) and cos(x), plot both curves on the same
# figure, and use a legend to distinguish them.

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)

y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")

plt.title("sin(x) and cos(x)")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()

plt.show()
```

```python
# Q30
# Approach: Create one dataset and display it using four different
# chart types in a 2x2 grid: line, bar, scatter, and histogram.

import numpy as np
import matplotlib.pyplot as plt


days = np.array([1, 2, 3, 4, 5])
marks = np.array([60, 75, 70, 85, 90])


fig, axes = plt.subplots(2, 2)

# 1. Line chart
axes[0, 0].plot(days, marks)
axes[0, 0].set_title("Line Chart")

# 2. Bar chart
axes[0, 1].bar(days, marks)
axes[0, 1].set_title("Bar Chart")

# 3. Scatter chart
axes[1, 0].scatter(days, marks)
axes[1, 0].set_title("Scatter Chart")

# 4. Histogram
axes[1, 1].hist(marks, bins=5)
axes[1, 1].set_title("Histogram")

plt.show()
```

```python
# Q32
# Approach: Create a DataFrame of 10 students and their scores
# in 3 subjects. Group the data by student, calculate each
# student's average score, sort the averages from highest to
# lowest, and get the top performer.

import pandas as pd

df = pd.DataFrame({
    "student": [
        "Ali", "Ali", "Ali",
        "Ahmed", "Ahmed", "Ahmed",
        "Ayesha", "Ayesha", "Ayesha",
        "Bilal", "Bilal", "Bilal",
        "Fatima", "Fatima", "Fatima",
        "Hamza", "Hamza", "Hamza",
        "Hira", "Hira", "Hira",
        "Sara", "Sara", "Sara",
        "Usman", "Usman", "Usman",
        "Zain", "Zain", "Zain"
    ],

    "subject": [
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics",
        "Math", "English", "Physics"
    ],

    "score": [
        80, 70, 90,
        70, 80, 78,
        90, 85, 90,
        80, 85, 85,
        95, 90, 92,
        75, 80, 85,
        90, 92, 93,
        85, 90, 95,
        70, 75, 80,
        85, 88, 90
    ]
})

student_average = df.groupby("student")["score"].mean()

sorted_average = student_average.sort_values(ascending=False)

top_student = sorted_average.index[0]
top_average = sorted_average.iloc[0]

print("Average score of each student:")
print(student_average)

print("\nTop performer:", top_student)
print("Top average:", top_average)
```

```python
# Q34
# Approach: Create a DataFrame with missing scores,
# calculate the mean of the score column, fill NaN values
# with that mean, and confirm that no NaNs remain.

import pandas as pd

df = pd.DataFrame({
    "student": ["Ali", "Ahmed", "Ayesha", "Bilal", "Fatima"],
    "score": [80, 70, None, 90, None]
})

mean_score = df["score"].mean()

df["score"] = df["score"].fillna(mean_score)

print(df)


print("Number of NaNs:", df["score"].isna().sum())
```
